## 1. Dane i model

```python
import random
from collections import Counter

import numpy as np
import torch
from datasets import load_dataset

SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)
```

```python
DATASET_NAME = "cornell-movie-review-data/rotten_tomatoes"
CHECKPOINT = "distilbert/distilbert-base-uncased"

dataset = load_dataset(DATASET_NAME)
label_names = dataset["train"].features["label"].names
id2label = dict(enumerate(label_names))
label2id = {label: label_id for label_id, label in id2label.items()}

print("Checkpoint:", CHECKPOINT)
print("Seed:", SEED)
print("Mapowanie klas:", id2label)
for split_name in ("train", "validation", "test"):
    split = dataset[split_name]
    counts = Counter(split["label"])
    class_counts = {label: counts[label_id] for label_id, label in id2label.items()}
    print(f"{split_name}: {len(split)} przykładów, {class_counts}")
```
